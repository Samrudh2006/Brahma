## Step 4: Load XTTS v2 Model onto GPU

```python
import torch
from TTS.api import TTS

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

# Load XTTS v2 (Zero-shot cloning model)
tts = TTS("tts_models/multilingual/multi-dataset/xtts_v2").to(device)
print("✅ XTTS v2 Model ready for cloning!")
```

## Step 5: Generate Audio in Her Exact Voice!
You can test Telugu, English, or Hindi text here.

```python
from IPython.display import Audio, display

# 1. English BBA / Finance Topic Explanation
en_text = "Welcome to Brahma. Today let us understand Working Capital Management. Working capital is the difference between a firm's current assets and its current liabilities, crucial for daily business operations."
out_en = "cloned_voice_english_finance.wav"
tts.tts_to_file(text=en_text, speaker_wav=audio_path, language="en", file_path=out_en)
print("🎧 English Cloned Audio:")
display(Audio(out_en))

# 2. Telugu Business / Finance Greeting
te_text = "నమస్కారం, బ్రహ్మ ఏఐ కి స్వాగతం. మీ వ్యాపార ఆర్థిక ప్రణాళిక మరియు పెట్టుబడి నిర్ణయాలలో నేను మీకు సహాయం చేస్తాను."
out_te = "cloned_voice_telugu_finance.wav"
tts.tts_to_file(text=te_text, speaker_wav=audio_path, language="hi", file_path=out_te)
print("🎧 Telugu Cloned Audio:")
display(Audio(out_te))

# 3. Hindi Greeting
hi_text = "नमस्ते, ब्रह्मा एआई में आपका स्वागत है। मैं आपकी वित्तीय योजना और व्यावसायिक रणनीति में मदद करूंगी।"
out_hi = "cloned_voice_hindi_finance.wav"
tts.tts_to_file(text=hi_text, speaker_wav=audio_path, language="hi", file_path=out_hi)
print("🎧 Hindi Cloned Audio:")
display(Audio(out_hi))
```

## Step 7 (Optional): Launch an Interactive Voice Clone Web UI (Gradio)
Run this cell to get a public link where you can type any text and hear her cloned voice instantly!

```python
!pip install -q gradio
import gradio as gr

def clone_speech(text, language):
    out_path = "output_cloned.wav"
    lang_code = "hi" if language in ["Telugu", "Hindi"] else "en"
    tts.tts_to_file(text=text, speaker_wav=audio_path, language=lang_code, file_path=out_path)
    return out_path

ui = gr.Interface(
    fn=clone_speech,
    inputs=[
        gr.Textbox(label="Script to Speak", value="Welcome to Brahma AI, the future of intelligence."),
        gr.Dropdown(choices=["English", "Telugu", "Hindi"], value="English", label="Language")
    ],
    outputs=gr.Audio(label="Her Cloned Voice Audio"),
    title="🔱 Brahma AI — Custom Voice Cloner",
    description="Speaks any text using the reference voice recording."
)

ui.launch(share=True)
```
